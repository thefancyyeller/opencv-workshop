# Program 1: Image Loading With OpenCV
The code below loads in your images, and shows you what images are recognized in the project. 

Not only does it do that, but it displays all of your images. This serves 2 purposes. First, convenience to the user. Second, it is a demonstration of how OpenCV as a library can be used to manipulate image files.

```python
import cv2 # OpenCV
import numpy as np
from IPython.display import Image, display # Specific to Jupyter Notebook for showing the images
import os, re

# User Options
show_images = True # Disable if too many images
images_per_row = 3
displayDims = [200, 200] # Size of scaled image preview

# Assembles a list of image filepaths into 1 singular image
def stitch(filenames: list[str]):
    # Helper vars
    out = None # This is the actual image file being shown to user, a grid of existing images
    buffer = [] # Holds a singular row of image data so we may accumulate multiple before assembling our row.

    # Open, resize, and merge all images
    for index, fname in enumerate(imFiles): # Open and scale the images
        scaled = cv2.imread(f"./images/{fname}")
        if scaled is not None: # Skip files OpenCV can't read
            buffer.append(cv2.resize(scaled, displayDims))
        # Add buffer to output image
        if buffer and (len(buffer) == images_per_row or index == (len(imFiles)-1)):
            while len(buffer) < images_per_row: # Pad the last row with blanks so widths match
                buffer.append(np.zeros_like(buffer[0]))
            row = cv2.hconcat(buffer)
            if out is None:
                out = row
            else:
                out = cv2.vconcat([out, row])
            buffer = []
    return out

# List the found image files to the user
imFiles = [x.name for x in os.scandir('images') if x.is_file() and re.match(".*\\.(jpe?g|png|webp|bmp)", x.name)]
print("Files found:")
for i in imFiles:
    print (i)

# Display the images in this notebook
out = stitch(imFiles)
# Display our assembled file if user allows and we actually have an image to show
if show_images and out is not None:
    ok, encoded = cv2.imencode('.png', out)
    display(Image(data=encoded.tobytes()))


```

# Program 2: Face Detection
OpenCV is helpful for basic image manipulation, as demonstrated above. This, however, is only a mere part of this library. The next feature we will look at will be facial recognition within OpenCV. OpenCV comes with the tools to build your own models, or use pre-existing ones. This excersize will download a pre-trained model and use it to label if each photo has a face or not within it.

```python
import urllib.request

model_path = "models/face_detection_yunet_2023mar.onnx"
if not os.path.exists(model_path):
    os.makedirs("models", exist_ok=True)
    urllib.request.urlretrieve(
        "https://github.com/opencv/opencv_zoo/raw/main/models/face_detection_yunet/face_detection_yunet_2023mar.onnx",
        model_path)

detector = cv2.FaceDetectorYN.create(model_path, "", (320, 320), score_threshold=0.8)

# Check every image for a face
for fname in imFiles:
    img = cv2.imread(f"./images/{fname}")
    if img is None: # Skip files OpenCV can't read
        print(f"{fname}: could not read file")
        continue
    scale = min(1.0, 640 / max(img.shape[:2])) # YuNet struggles with huge photos, so shrink anything over 640px
    img = cv2.resize(img, None, fx=scale, fy=scale)
    detector.setInputSize((img.shape[1], img.shape[0])) # Detector must match the image's (width, height)
    _, faces = detector.detect(img) # None when no face is found
    has_face = faces is not None
    print(f"{fname}: {has_face}")
```

# Program 3: Face Tracking

To graduate a step from the previous example, we will now use this same pre-downloaded model but draw a box around the found faces.


```python
import urllib.request

# YuNet is OpenCV's modern face detector. It needs a small pre-trained model file, downloaded once.
model_path = "models/face_detection_yunet_2023mar.onnx"
if not os.path.exists(model_path):
    os.makedirs("models", exist_ok=True)
    urllib.request.urlretrieve(
        "https://github.com/opencv/opencv_zoo/raw/main/models/face_detection_yunet/face_detection_yunet_2023mar.onnx",
        model_path)

img = out.copy() # We will use the stitched image from Program 1 (copied so the original stays clean)
h, w = img.shape[:2]
detector = cv2.FaceDetectorYN.create(model_path, "", (w, h), score_threshold=0.8)

_, faces = detector.detect(img) # One row per face: x, y, w, h, 5 landmarks, confidence score
if faces is None:
    faces = []

for face in faces:
    x, y, fw, fh = face[:4].astype(int)
    cv2.rectangle(img, (x, y), (x + fw, y + fh), (0, 255, 0), 2) # green box, 2px thick

print(f"Found {len(faces)} face(s)")
ok, encoded = cv2.imencode('.png', img)
display(Image(data=encoded.tobytes()))
```

# Program 4: Face Comparison
Finally, detecting a face has applications, however differentiating them is a task onto itself. This final program duplicates all images and mirrors them (different pixel representations of the same information), finds each face, and tags them with a fingerprint, plus similarity score if they are not the first to be given that fingerprint.

The bar along the bottom of each box shows how similar that face is to its closest match (`vs #3` means face #3), as a percentage of the match threshold. The white tick marks 100%: a green bar past it means the same person, an orange bar short of it means a different person.

```python
import urllib.request

# Two pre-trained models: YuNet finds faces, SFace turns each face into a 128-number "fingerprint". Downloaded once.
models = {
    "models/face_detection_yunet_2023mar.onnx": "https://github.com/opencv/opencv_zoo/raw/main/models/face_detection_yunet/face_detection_yunet_2023mar.onnx",
    "models/face_recognition_sface_2021dec.onnx": "https://github.com/opencv/opencv_zoo/raw/main/models/face_recognition_sface/face_recognition_sface_2021dec.onnx",
}
for path, url in models.items():
    if not os.path.exists(path):
        os.makedirs("models", exist_ok=True)
        urllib.request.urlretrieve(url, path)

# Build the test image: the grid from Program 1, with a copy underneath where every photo is mirrored left-to-right
tile_w, tile_h = displayDims
mirrored = out.copy()
for y in range(0, out.shape[0], tile_h):
    for x in range(0, out.shape[1], tile_w):
        mirrored[y:y + tile_h, x:x + tile_w] = cv2.flip(out[y:y + tile_h, x:x + tile_w], 1) # 1 = flip horizontally
img = cv2.vconcat([out, mirrored])
top_h = out.shape[0] # Anything above this line is an original, below it is a mirrored copy

# Find every face
h, w = img.shape[:2]
detector = cv2.FaceDetectorYN.create("models/face_detection_yunet_2023mar.onnx", "", (w, h), score_threshold=0.8)
_, faces = detector.detect(img)
if faces is None:
    faces = []

# Fingerprint every face: line it up using its landmarks, then let SFace describe it as 128 numbers
recognizer = cv2.FaceRecognizerSF.create("models/face_recognition_sface_2021dec.onnx", "")
originals = [] # (face, fingerprint, filename) for faces in the top half
copies = [] # (face, fingerprint, filename) for faces in the bottom half
for face in sorted(faces, key=lambda f: (int(f[1] // tile_h), f[0])): # Reading order: row by row, left to right
    fingerprint = recognizer.feature(recognizer.alignCrop(img, face))
    row, col = int((face[1] + face[3] / 2) // tile_h), int((face[0] + face[2] / 2) // tile_w) # Which grid tile the face is in
    if face[1] + face[3] / 2 < top_h:
        originals.append((face, fingerprint, imFiles[row * images_per_row + col]))
    else:
        copies.append((face, fingerprint, imFiles[(row - top_h // tile_h) * images_per_row + col]))

match_threshold = 0.363 # SFace's recommended cosine-similarity cutoff for "same person"

def similarity(fp_a, fp_b):
    return recognizer.match(fp_a, fp_b, cv2.FaceRecognizerSF_FR_COSINE)

# Draw on a 2x enlarged copy so the labels are readable
draw_scale = 2
canvas = cv2.resize(img, None, fx=draw_scale, fy=draw_scale)

# Box with an ID strip on top, a similarity meter along the bottom, and the closest match written underneath.
# The meter runs from 0 to twice the threshold, so the white tick in the middle is the threshold:
# past the tick (100%+) means "same person".
def draw_face(face, text, closest_id, score, color):
    x, y, fw, fh = (face[:4] * draw_scale).astype(int)
    cv2.rectangle(canvas, (x, y), (x + fw, y + fh), color, 2)
    cv2.rectangle(canvas, (x, y), (x + fw, y + 24), color, -1) # Solid strip along the top of the box so the text is readable
    cv2.putText(canvas, text, (x + 4, y + 18), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 0, 0), 2) # Black text inside the strip

    percent = score / match_threshold * 100
    bar_top = y + fh - 10
    fill = int(fw * min(score / (2 * match_threshold), 1))
    bar_color = (0, 255, 0) if score >= match_threshold else (0, 165, 255) # Green above the threshold, orange below
    cv2.rectangle(canvas, (x, bar_top), (x + fw, y + fh), (40, 40, 40), -1) # Dark background
    cv2.rectangle(canvas, (x, bar_top), (x + fill, y + fh), bar_color, -1) # Filled part
    cv2.line(canvas, (x + fw // 2, bar_top), (x + fw // 2, y + fh), (255, 255, 255), 2) # Threshold tick
    label = f"vs #{closest_id}: {percent:.0f}%"
    (text_w, _), _ = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, 0.55, 1)
    cv2.rectangle(canvas, (x, y + fh), (x + max(fw, text_w + 8), y + fh + 22), (40, 40, 40), -1) # Dark strip under the box
    cv2.putText(canvas, label, (x + 4, y + fh + 16), cv2.FONT_HERSHEY_SIMPLEX, 0.55, (255, 255, 255), 1) # White text on it

# Top half: a face that matches someone already seen reuses their ID, otherwise it gets the next new ID
ids = [] # ID given to each original, in the same order as originals
next_id = 1
for i, (face, fingerprint, fname) in enumerate(originals):
    earlier = [(similarity(fingerprint, originals[j][1]), ids[j]) for j in range(i)]
    score, face_id = max(earlier) if earlier else (0, None)
    if score < match_threshold: # Nobody earlier is close enough, so this is a new person
        face_id = next_id
        next_id += 1
    ids.append(face_id)

# Find the most similar *other* original for each face, so the meter shows something for every photo
print("Original photos:")
for i, (face, fingerprint, fname) in enumerate(originals):
    others = [(similarity(fingerprint, originals[j][1]), j) for j in range(len(originals)) if j != i]
    score, j = max(others) if others else (0, i)
    draw_face(face, f"#{ids[i]}", ids[j], score, (0, 255, 0))
    print(f"  {fname} (#{ids[i]}): most similar to {originals[j][2]} (#{ids[j]}), score {score:.2f} = {score / match_threshold:.0%} of threshold")

# Bottom half: compare each mirrored face with every original and keep the closest match
print("Mirrored copies:")
for face, fingerprint, fname in copies:
    score, j = max((similarity(fingerprint, orig_fp), j) for j, (_, orig_fp, _) in enumerate(originals))
    if score >= match_threshold:
        draw_face(face, f"#{ids[j]}", ids[j], score, (0, 255, 0)) # Green: matched an original
    else:
        draw_face(face, "?", ids[j], score, (0, 0, 255)) # Red: no original was close enough
    print(f"  {fname} (mirrored): most similar to {originals[j][2]} (#{ids[j]}), score {score:.2f} = {score / match_threshold:.0%} of threshold")

ok, encoded = cv2.imencode('.png', canvas)
display(Image(data=encoded.tobytes()))
```
